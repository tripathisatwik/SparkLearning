#Learning PySpark


```python
import pyspark
from pyspark.sql import SparkSession
```

```python
#Creating Spark Session
spark = SparkSession.builder.appName('LearningPySpark').getOrCreate()
```

```python
# Loading csv dataset, with header and schema
# header=True: if dataset has column name
# inferSchema=True: if we want to use dataset schema or all dtypes are infered as string
df = spark.read.csv("/content/Mall_Customers.csv",header=True,inferSchema=True)
```

```python
df.show()
```

```python
#Checking the schema of spark dataframe
df.printSchema()
```

```python
# Showing the type of dataframe data structure
type(df)
```

```python
#listing the column names
df.columns
```

```python
#listing datatypes of dataset
df.dtypes
```

```python
#finding baisc stats using describe()
df.describe().show()
```

```python
from IPython.core.interactiveshell import dis
#selecting a single column
display(df.select('Gender').head(5))

#selecting multiple columns
display(
    df.select(['Gender','Age'])
    .show()
)
```

```python
#Adding New columns
df=df.withColumn("Age After 5 Years", df['Age']+5)
df.show()
```

```python
#dropping columns
df=df.drop('Age After 5 Years')
df.show()
```

```python
#Renaming Columns
df=df.withColumnRenamed('Annual Income (k$)','Yearly Income (k$)')
df.show()
```

```python
#dropping columns
df.drop('Gender').show()
```

```python
#dropping null values
df.na.drop().show() #will drop all rows with null values
df.na.drop(how='all').show() #will only drop if all value in row is null
df.na.drop(how='any').show() #will only drop if any value in row is null
df.na.drop(how='any',thresh=3).show() # if row contains more than 3 null it gets deleted
df.na.drop(how='any',subset=['Yearly Income (k$)']).show() #deletes rows only from the specified column
```

```python
#filling the missing values
df.na.fill('Missing Value',['Age','Gender']).show()
```

```python
#imputed function to replace null value with mean
from pyspark.ml.feature import Imputer

imputer = Imputer (
    inputCols=['Age','Yearly Income (k$)','Spending Score (1-100)'],
    outputCols=['{}_imputed'.format(c) for c in ['Age','Yearly Income (k$)','Spending Score (1-100)']]
).setStrategy('mean')
```

```python
imputer.fit(df).transform(df).show()
```

## Filter Operations in Spark

```python
#Renaming the column for easier use
df=df.withColumnRenamed("Yearly Income (k$)","AnnualIncome")
df.show()
```

```python
# finding datas with AnnualIncome less then 15,000
df.filter("AnnualIncome<=15").show()
```

```python
# finding gender,age,annualincome with AnnualIncome more then 100,000
df.filter("AnnualIncome>=100").select(['Gender','Age','AnnualIncome']).show()
```

```python
# finding datas with AnnualIncome more then 100,000 and age greater than 30
df.filter((df['AnnualIncome']>=100) & (df['Age']<=30)).show()
```

```python
# finding datas with AnnualIncome more then 100,000 and age not greater than 30
df.filter((df['AnnualIncome']>=100) & ~(df['Age']<=30)).show()
```

##GroupBy and Aggregate Functions

```python
# groupby to max salary of customer based on id
df.groupBy('CustomerID').max('AnnualIncome').show()
```

```python
# groupby to sum of salary of customer based on gender
df.groupBy('Gender').max('AnnualIncome').show()
```

```python
# groupby to mean of salary of customer based on gender
df.groupBy('Gender').mean('AnnualIncome').show()
```

```python
# groupby to sum of salary of customer based on gender
df.groupBy('Gender').count().show()
```

```python
# groupby to find sepnding score of customer based on gender
df.groupBy('Gender').max('Spending Score (1-100)').show()
```
